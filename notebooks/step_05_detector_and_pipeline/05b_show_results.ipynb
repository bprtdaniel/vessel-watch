## How far did detector training get?

```python
from pathlib import Path
run = Path('/content/drive/MyDrive/5.Projects/runs/yolo11s_obb')
rows = (run / 'results.csv').read_text().strip().splitlines()
print('epochs logged:', len(rows) - 1, 'of 100')
print('weights:', sorted(p.name for p in (run / 'weights').iterdir()))
print('test metrics written:', (run / 'metrics_test.json').exists())
print('pipeline results:', sorted(p.name for p in run.parent.glob('pipeline_*.json')))
```

## Results

```python
import json
RUNS = Path('/content/drive/MyDrive/5.Projects/runs')

det = RUNS / 'yolo11s_obb' / 'metrics_test.json'
print('DETECTOR (test):', json.loads(det.read_text()) if det.exists() else 'metrics_test.json missing')

files = sorted(RUNS.glob('pipeline_*.json'))
print()
print(len(files), 'pipeline result files')
for f in files:
    r = json.loads(f.read_text())
    print()
    print(f"{r['name']} | level {r['level']} | crop {r['crop']} | conf {r['conf']}")
    for key in ('vessels', 'detections', 'detection_recall', 'detection_precision',
                'classification_accuracy_on_found', 'end_to_end_recall',
                'classifier_accuracy_on_true_boxes'):
        print(f'  {key:<36} {r[key]}')
```
