# vessel-watch: inventory of the saved models

Identifies every weight and history file in the Drive models folder from its contents, so messy file names do not matter. Read-only: it lists the folder and writes one `inventory.json`. No GPU needed.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
REPO_URL = 'https://github.com/bprtdaniel/vessel-watch.git'
MODELS_DIR = '/content/drive/MyDrive/5.Projects/models'
```

```python
import os, subprocess

REPO_DIR = '/content/vessel-watch'
if os.path.exists(REPO_DIR):
    subprocess.run(['git', '-C', REPO_DIR, 'pull'], check=True)
else:
    subprocess.run(['git', 'clone', REPO_URL, REPO_DIR], check=True)
%pip install -q --no-deps -e /content/vessel-watch
```
