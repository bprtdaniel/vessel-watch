# Step 7d: how far has the 10 m retraining got?

Reads the run folders of step 7c on Drive and prints, for each of the nine classifier runs, either its test result or how many epochs it has done, and the same for the detector. Useful when the Colab output of 7c stops updating or the session disconnects: run it twice a few minutes apart to see whether anything is still moving. No GPU needed, nothing is written.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import json
from pathlib import Path
RUNS = Path('/content/drive/MyDrive/5.Projects/runs')

for level in (1, 2, 3):
    for model in ('net', 'net_max', 'resnet101'):
        run = RUNS / f'crops10m_level{level}_{model}'
        history = run / 'history.json'
        epochs = len(json.loads(history.read_text())['val_accuracies']) if history.exists() else 0
        if (run / 'metrics_test.json').exists():
            m = json.loads((run / 'metrics_test.json').read_text())
            print(f"{run.name:<28} finished | test acc {m['accuracy']:.1f} | macro-F1 {m['macro_f1']:.1f}")
        else:
            print(f"{run.name:<28} {epochs}/30 epochs")

det = RUNS / 'yolo11s_obb_10m'
rows = (det / 'results.csv').read_text().strip().splitlines() if (det / 'results.csv').exists() else []
print('detector at 10 m:', 'finished' if (det / 'metrics_test.json').exists() else f'{max(0, len(rows) - 1)}/100 epochs')
```
