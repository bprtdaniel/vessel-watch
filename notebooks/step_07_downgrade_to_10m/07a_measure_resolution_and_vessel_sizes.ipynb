# Step 7a: what is left of ShipRSImageNet at 10 m?

First part of step B of the Sentinel-2 experiment. Before the dataset is resampled to Sentinel-2's 10 m and the models are retrained, this measures three things that decide how that is done:

1. Which ground resolution each image has. The dataset mixes sources between about 0.12 and 6 m per pixel.
2. How small the scenes become at 10 m.
3. How long the vessels are in metres, and so how many of them are still more than a pixel or two.

Reads the annotation files only. No GPU needed, runs in a minute or two, nothing is written.
