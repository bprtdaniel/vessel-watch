# Step 4c: per-class and per-size results of the classifier study

Reads the `metrics_test.json` files that step 4a wrote to Drive and prints, for the ResNet101 at levels 1 and 3, the accuracy by vessel size, the accuracy without the `Dock` class, and precision, recall and F1 for every class. No GPU needed, nothing is written.

Run on 2026-10-09.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import json
from pathlib import Path
RUNS = Path('/content/drive/MyDrive/5.Projects/runs')
print('runs folder exists:', RUNS.exists())
print(sorted(p.name for p in RUNS.iterdir()) if RUNS.exists() else '')

for run in ('crops_level1_resnet101', 'crops_level3_resnet101'):
    f = RUNS / run / 'metrics_test.json'
    print()
    if not f.exists():
        print(f'{run}: missing. Folder holds:', sorted(p.name for p in f.parent.iterdir()) if f.parent.exists() else 'no folder')
        continue
    m = json.loads(f.read_text())
    print(f"=== {run} ===  by size: {m['by_size']}")
    print('vessels only:', m.get('vessels_only'))
    for c in sorted(m['per_class'], key=lambda c: -c['support']):
        if c['support']:
            print(f"{c['class']:<28} n={c['support']:>4}  P={c['precision']:5.1f}  R={c['recall']:5.1f}  F1={c['f1']:5.1f}")
```
