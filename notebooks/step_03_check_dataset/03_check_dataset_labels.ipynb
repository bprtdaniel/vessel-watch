# Step 3: what labels does the dataset have?

Two questions that decided how the corrected study was built: does the dataset's test split have labels, and does it store oriented (rotated) boxes? Reads the annotation files on Drive and prints what it finds. No GPU needed, nothing is written.

Run on 2026-10-08. Answer: the test split has no labels, and every annotation carries the four corners of an oriented box in `segmentation`.

```python
import json, itertools
from pathlib import Path

ROOT = Path('/content/drive/MyDrive/5.Projects/Data/ShipRSImageNet_V1')

print('Folders:', sorted(p.name for p in ROOT.iterdir()))
for sub in ('COCO_Format', 'VOC_Format'):
    if (ROOT / sub).exists():
        print(f'{sub}:', sorted(p.name for p in (ROOT / sub).iterdir())[:30])

# Check 1: does the test split have labels?
print()
print('--- Check 1: annotations per split ---')
for split in ('train', 'val', 'test'):
    f = ROOT / 'COCO_Format' / f'ShipRSImageNet_bbox_{split}_level_1.json'
    if not f.exists():
        print(split, 'FILE MISSING'); continue
    d = json.loads(f.read_text())
    anns = d.get('annotations', [])
    print(f"{split}: {len(d['images'])} images, {len(anns)} annotations")

# Check 2: are there oriented boxes?
print()
print('--- Check 2a: one COCO annotation ---')
d = json.loads((ROOT / 'COCO_Format' / 'ShipRSImageNet_bbox_train_level_1.json').read_text())
print(json.dumps(d['annotations'][0], indent=1)[:800])

print()
print('--- Check 2b: one VOC XML file ---')
xml = next(itertools.chain((ROOT / 'VOC_Format').rglob('*.xml')), None)
print(xml)
if xml:
    print(xml.read_text()[:2500])
```
